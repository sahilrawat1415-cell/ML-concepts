```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Converting emotions colum to numeric


```python
unique_emo=df['emotion'].unique()
emo_number={}
i=0
for emo in unique_emo:
    emo_number[emo]= i
    i+=1
df['emotion']=df['emotion'].map(emo_number)
```

```python
df
```

Checking if text is lower or not


```python
df['text']=df['text'].apply(lambda x:x.lower())
```

```python
df
```

Remove Punctuation

```python
import string
def remove_punc(txt):
    return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text']=df['text'].apply(remove_punc)
```

```python
df
```

Remove Numbers

```python
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new

df['text'] = df['text'].apply(remove_numbers)
```

Remove URl

```python
import re

def remove_urls(txt):
    return re.sub(r'https?://\S+|www\.\S+', '', txt)

df['text'] = df['text'].apply(remove_urls)
```

Remove Emojis

```python
def remove_emoji(txt):
    new=""
    for i in txt:
        if i.isascii():
            new +=i
    return new
df['text'] = df['text'].apply(remove_emoji)
```

Tokenization

```python
def remove(txt):
    words=word_tokenize(txt)
    cleaned=[]
    
    for i in words:
        if not i in stop_words:
            cleaned.append(i)
            
    return ' '.join(cleaned)
```

```python
df['text'] = df['text'].apply(remove)
```

Bag of Words


```python
from sklearn.feature_extraction.text import CountVectorizer


documents=[
    "I love pizza",
    "Pizza is the best",
    "I love pasta",
    "Pasta is great"
]

vectorizer=CountVectorizer()

X=vectorizer.fit_transform(documents)
print("Vocabulary :",vectorizer.get_feature_names_out())
print("Bow Matrix : \n",X.toarray())
```

N_gram range

```python
from sklearn.feature_extraction.text import CountVectorizer


documents=[
    "I love pizza",
    "Pizza is the best",
    "I love pasta",
    "Pasta is great"
]

vectorizer=CountVectorizer(ngram_range=(3,3))

X=vectorizer.fit_transform(documents)
print("Vocabulary :",vectorizer.get_feature_names_out())
print("Bow Matrix : \n",X.toarray())
```

TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer


documents=[
    "I love pizza",
    "Pizza is the best",
    "I love pasta",
    "Pasta is great"
]

vectorizer=TfidfVectorizer()

X=vectorizer.fit_transform(documents)
print("Vocabulary :",vectorizer.get_feature_names_out())
print("Bow Matrix : \n",X.toarray())
```
