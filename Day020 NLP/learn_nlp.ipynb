Converting emotions colum to numeric


```python
unique_emo=df['emotion'].unique()
emo_number={}
i=0
for emo in unique_emo:
    emo_number[emo]= i
    i+=1
df['emotion']=df['emotion'].map(emo_number)
```

```python
df
```

Checking if text is lower or not


```python
df['text']=df['text'].apply(lambda x:x.lower())
```

```python
df
```

Remove Punctuation

```python
import string
def remove_punc(txt):
    return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text']=df['text'].apply(remove_punc)
```

```python
df
```

Remove Numbers

```python
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new

df['text'] = df['text'].apply(remove_numbers)
```

Remove URl

```python
import re

def remove_urls(txt):
    return re.sub(r'https?://\S+|www\.\S+', '', txt)

df['text'] = df['text'].apply(remove_urls)
```

Remove Emojis

```python
def remove_emoji(txt):
    new=""
    for i in txt:
        if i.isascii():
            new +=i
    return new
df['text'] = df['text'].apply(remove_emoji)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
stop_words=set(stopwords.words("english"))
len(stop_words)
```

```python
df.loc[1],['text']
```

Tokenization

```python
def remove(txt):
    words=word_tokenize(txt)
    cleaned=[]
    
    for i in words:
        if not i in stop_words:
            cleaned.append(i)
            
    return ' '.join(cleaned)
```

```python
df['text'] = df['text'].apply(remove)
```
